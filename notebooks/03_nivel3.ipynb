# Preparação
Cada notebook roda sozinho: estas células refazem o `df`, o `df_receitas` e o `df_fundao` dos exercícios anteriores (mesmo código dos exercícios 1, 3, 5 e 6).

```python
import pandas as pd

df = pd.read_csv("../data/receitas_candidatos_2026_RS.csv", sep=";", decimal=",", encoding="latin-1")

colunas = ["NM_CANDIDATO",
           "SG_PARTIDO",
           "DS_CARGO",
           "DS_GENERO",
           "DS_FONTE_RECEITA",
           "DS_ORIGEM_RECEITA",
           "DT_RECEITA",
           "VR_RECEITA"
           ]

df_receitas = df[colunas]

filtro_fundao_eleitoral = df_receitas["DS_FONTE_RECEITA"] == "FUNDO ESPECIAL"
df_fundao = df_receitas[filtro_fundao_eleitoral]
```

# Exercício 12

```python
filtro_fudao_dfederal = df_fundao["DS_CARGO"] == "Deputado Federal"

df_fundao_dfederal=df_fundao[filtro_fudao_dfederal]

maior_valor = df_fundao_dfederal["VR_RECEITA"].max()
filtro_maior_valor = df_fundao_dfederal["VR_RECEITA"] == maior_valor
df_maior_valor = df_fundao_dfederal[filtro_maior_valor]

nomes = ", ".join(df_maior_valor["NM_CANDIDATO"].unique())

if df_maior_valor["NM_CANDIDATO"].nunique() == 1:
    print(f"O candidato(a) com maior valor foi {df_maior_valor["NM_CANDIDATO"].iloc[0]} com R$ {df_maior_valor["VR_RECEITA"].iloc[0]:,.2f}")
else:
    print(f'Os candidatos(as) com maior transferência individual do fundão são: {nomes}\n No total são {df_maior_valor["NM_CANDIDATO"].nunique()} candidatos com valor {df_maior_valor["VR_RECEITA"].iloc[0]:,.2f} cada.')
```

# Exercício 13
Escolha três partidos que você acompanha. Quanto cada um recebeu do fundão no RS? Qual dos três recebeu mais?

```python
# lista_partidos = []
# n_partidos = 3
# for i in range(n_partidos):
#     partido = input("Digite o nome do partido: ")
#     lista_partidos.append(partido.upper())

lista_partidos = ['PL', 'PT', 'PSOL']
maior_valor = 0
maior_partido = ""

for partido in lista_partidos:
    filtro_partido = df_fundao["SG_PARTIDO"] == partido
    valor = df_fundao[filtro_partido]["VR_RECEITA"].sum()
    if valor > maior_valor:
        maior_valor = valor
        maior_partido = partido
    print(f"O partido {partido} recebeu: R$ {valor:,.2f}")

print(f"O partido que mais recebeu fundão eleitoral foi o {maior_partido} que recebeu R$ {maior_valor:,.2f}")
    
```

# Exercício 14
Em qual cargo o fundão pesa mais: qual cargo tem a maior parcela do seu dinheiro vinda do fundão?

```python
fundao_por_cargo = df_fundao.groupby("DS_CARGO")["VR_RECEITA"].sum()
tota_por_cargo = df_receitas.groupby("DS_CARGO")["VR_RECEITA"].sum()

percentual_por_cargo = (fundao_por_cargo / tota_por_cargo) * 100

cargo_vencedor = percentual_por_cargo.idxmax()
maior_percentual = percentual_por_cargo.max()

print(f"O cargo em que o fundão pesa mais é {cargo_vencedor}, com {maior_percentual:.1f}% do dinheiro vindo do fundão.")
```
